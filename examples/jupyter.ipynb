# MathCAS v2 in Jupyter

Each code cell runs in one MathCAS session; the value of the last statement is shown below the cell. Select the **MathCAS v2** kernel.

```mathcas
1/3 + 1/6
```

## Exact, machine and arbitrary precision

`n(x)` gives a machine real, `n(x, d)` a ball with at least `d` correct digits.

```mathcas
q = 1/3 + 2/7
[n(q), n(q, 40)]
```

```mathcas
n(pi, 50)
```

```mathcas
b = 1~20
(b + 10^-15) - b
```

## Roots and minima

```mathcas
find_root(cos(x) == x, [x, 1], digits: 40)
```

```mathcas
find_minimum((x - 1)^2 + 100(y - x^2)^2, [[x, -1], [y, 1]])
```

A solver that does not converge returns a `failure` value; the session continues.

```mathcas
find_root(exp(x) + 1, [x, 0])
```

## Compiled loops

```mathcas
@compiled fn logistic(r: Real, n: Int) = {
  var x = 0.5
  for k in range(n) { x = r*x*(1.0 - x) }
  x
}
map(r => logistic(r, 1000), [2.8, 3.2, 3.5, 3.7])
```

## Printing

`say` prints from inside a cell.

```mathcas
for k in range(3) { say("k = ", k, ", k! = ", factorial(k)) }
```
